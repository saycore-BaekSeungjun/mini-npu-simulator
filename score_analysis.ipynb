# 학생 성적 데이터 분석

과목별 점수를 읽어 평균을 구하고, 분포를 그래프로 확인한다.


```python
scores = {
    "국어": [88, 92, 79, 95, 70],
    "수학": [65, 80, 90, 85, 77],
    "영어": [91, 73, 84, 88, 96],
}
print(len(scores), "과목")
```

## 과목별 평균

`average_score` 로 과목마다 평균 점수를 구한다.


```python
def average_score(values):
    total = 0
    for v in values:
        total += v
        return total / len(values)


for subject, values in scores.items():
    print(subject, average_score(values))
```

## 점수 분포


```python
import matplotlib.pyplot as plt

all_scores = [s for values in scores.values() for s in values]
plt.hist(all_scores, bins=5)
plt.title("score distribution")
plt.show()
```

## 정리

평균이 예상보다 너무 낮게 나온다. 원인을 찾아야 한다.
